```python
from openai import OpenAI

# 클라이언트 생성 (환경 변수 OPENAI_API_KEY를 자동으로 읽어옵니다)
# client = OpenAI()

# 또는 API 키를 코드 내에 직접 전달할 경우:
client = OpenAI(api_key="")

# Chat Completion 요청
response = client.chat.completions.create(
    model="gpt-4o-mini",  # 빠르고 저렴한 gpt-4o-mini 모델 추천 (필요 시 gpt-4o 사용)
    messages=[
        {"role": "system", "content": "너는 친절한 AI 도우미야."},
        {"role": "user", "content": "안녕 나는 홍길동이야."}
    ]
)

# 결과 출력
print(response.choices[0].message.content)
```
